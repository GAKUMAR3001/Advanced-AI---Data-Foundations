```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/lec9.csv')
```

```python
df.shape
```

```python
df.head()
```

```python
df[['month_year', 'rating_filled']]
```

```python
df[['month_year', 'rating_filled']]
```

```python
monthly_rating = df.groupby(['month_year'])['rating_filled'].mean().reset_index()
```

```python
monthly_rating
```

```python
monthly_rating['month_year'] = pd.to_datetime(monthly_rating['month_year'],format = '%B %Y')
```

```python
monthly_rating = monthly_rating.sort_values(by = 'month_year')
```

```python
plt.figure(figsize=(10, 6))
plt.plot(monthly_rating['month_year'], monthly_rating['rating_filled'], marker ='o', color ='navy' )
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data = monthly_rating , x = 'month_year', y= 'rating_filled', marker ='o', color ='navy' )
plt.title("Rating over the month")
plt.grid(True, linestyle = '--')
plt.ylabel("Avg Rating", fontsize =12)
plt.xlabel('Month year',fontsize =12)
```

```python
plt.figure(figsize=(10, 6))
plt.bar(monthly_rating['month_year'], monthly_rating['rating_filled'], color ='navy' , width= 20)
plt.title("Rating over the month")

plt.ylabel("Avg Rating", fontsize =12)
plt.xlabel('Month year',fontsize =12)
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data = monthly_rating , x = 'month_year', y= 'rating_filled', color ='navy' , width= 0.8)
plt.title("Rating over the month")

plt.ylabel("Avg Rating", fontsize =12)
plt.xlabel('Month year',fontsize =12)
plt.xticks(rotation=45);
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data =df, y= 'rating')
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data =df, x = 'category',y= 'rating', palette='crest')
plt.xticks(rotation=45)
plt.title("Rating Distribution by Category", fontsize=16)
plt.xlabel("Category", fontsize=12)
plt.ylabel("Rating", fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.violinplot(data=df, x='rating_category', y='discounted_price', inner='quartile', palette='Accent')

plt.title("Violin Plot of Discounted Price by Rating Category", fontsize=16)
plt.xlabel("Rating Category", fontsize=12)
plt.ylabel("Discounted Price", fontsize=12)
plt.show()
```
